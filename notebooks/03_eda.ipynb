# Phase 5 — Exploratory data analysis

**Input:** `data/clean.parquet` · **Output:** 15 charts in `reports/figures/`, each followed by its insight.

Reminder: Severity = **impact on traffic (delay)**, not injuries. "High impact" below means Severity 3–4.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import seaborn as sns

FIGS = Path("../reports/figures")
FIGS.mkdir(parents=True, exist_ok=True)
SEV_COLORS = ["#9ecae1", "#4292c6", "#fd8d3c", "#a50f15"]  # Severity 1..4, light -> dark red
sns.set_theme(style="whitegrid", font_scale=0.95)

df = pd.read_parquet("../data/clean.parquet")
t = df["Start_Time"]  # local time of the accident
df = df.assign(hour=t.dt.hour, weekday=t.dt.dayofweek, month=t.dt.month, year=t.dt.year,
               high_impact=df["Severity"].ge(3))
print(df.shape)


def save(fig, name):
    fig.tight_layout()
    fig.savefig(FIGS / f"{name}.png", dpi=120, bbox_inches="tight")


def severity_share(by):
    """% of each Severity inside each group of `by` (rows sum to 100)."""
    return pd.crosstab(by, df["Severity"], normalize="index").mul(100)


def stacked(share, ax, horizontal=False):
    share.plot(kind="barh" if horizontal else "bar", stacked=True, color=SEV_COLORS, width=0.85, ax=ax)
    (ax.xaxis if horizontal else ax.yaxis).set_major_formatter(mtick.PercentFormatter())
    ax.legend(title="Severity", bbox_to_anchor=(1.01, 1), loc="upper left")
```

## 7b. Do the RQ1–RQ3 effects hold inside each data provider?

The providers label Severity very differently (chart 12): Source1 calls about 8% of its accidents high impact, Source2 about 34%. If a condition is more common in one provider's data, the pooled charts above can show an effect that is really a provider effect. So we repeat each comparison **inside each provider** (stratification).

For each condition: share of Severity 3–4 *with* the condition ÷ share *without* it (ratio > 1 = more high-impact accidents with the condition), and Cramér's V inside the provider.

```python
import sys
sys.path.insert(0, "..")
from src.features import HIGHWAY, BAD_WEATHER
import numpy as np
from scipy import stats

df = df.assign(
    is_highway=df["Street"].str.contains(HIGHWAY, regex=True, na=False).astype("int8"),
    is_night=df["Sunrise_Sunset"].astype(str).eq("Night").astype("int8"),
    bad_weather=df["Weather_Group"].astype(str).isin(BAD_WEATHER).astype("int8"),
)


def cramers_v(a, b):
    """Cramér's V between two categorical variables (no bias correction)."""
    ct = pd.crosstab(a, b)
    chi2 = stats.chi2_contingency(ct)[0]
    return np.sqrt(chi2 / (ct.to_numpy().sum() * (min(ct.shape) - 1)))


CONDITIONS = {"Night (RQ1)": "is_night", "Bad weather (RQ2)": "bad_weather", "Highway (RQ3)": "is_highway",
              "Junction (RQ3)": "Junction", "Traffic signal (RQ3)": "Traffic_Signal"}
rows = []
for src_name, g in [("All", df)] + [(s, df[df["Source"] == s]) for s in ["Source1", "Source2", "Source3"]]:
    for label, col in CONDITIONS.items():
        yes, no = g.loc[g[col] == 1, "high_impact"].mean() * 100, g.loc[g[col] == 0, "high_impact"].mean() * 100
        rows.append({"source": src_name, "condition": label, "rows": len(g), "sev3_4_%_with": yes,
                     "sev3_4_%_without": no, "ratio": yes / no, "cramers_v": cramers_v(g[col], g["Severity"])})
strat = pd.DataFrame(rows)
strat.pivot(index="condition", columns="source", values="ratio").loc[list(CONDITIONS)].round(2)
```

```python
# Severity 4 share at night vs day inside each provider (RQ1)
df.groupby(["Source", "is_night"], observed=True)["Severity"].apply(lambda s: s.eq(4).mean() * 100).unstack().round(2)
```

```python
STY = {"All": ("#222222", "D", 70, "All providers (pooled)"), "Source1": ("#4292c6", "o", 60, "Source1"),
       "Source2": ("#fd8d3c", "o", 60, "Source2"), "Source3": ("#74c476", "s", 35, "Source3 (1% of rows)")}
OFF = {"All": -0.24, "Source1": -0.08, "Source2": 0.08, "Source3": 0.24}
fig, ax = plt.subplots(figsize=(11, 4.8))
for s, (c, m, size, label) in STY.items():
    d = strat[strat["source"] == s].set_index("condition").loc[list(CONDITIONS)]
    y = np.arange(len(CONDITIONS)) + OFF[s]
    ax.scatter(d["ratio"], y, color=c, marker=m, s=size, zorder=3, label=label)
    for yy, r in zip(y, d["ratio"]):
        ax.plot([1, r], [yy, yy], color=c, lw=1.2, alpha=0.6, zorder=2)
        ax.text(r * (1.04 if r >= 1 else 0.96), yy, f"{r:.2f}", va="center", ha="left" if r >= 1 else "right",
                fontsize=7.5, color=c)
ax.axvline(1, color="k", lw=1)
ax.set_xscale("log")
ax.set_xticks([0.2, 0.33, 0.5, 1, 2, 3, 4], ["0.2", "0.33", "0.5", "1", "2", "3", "4"])
ax.xaxis.set_minor_formatter(mtick.NullFormatter())
ax.set_xlim(0.17, 5)
ax.set_yticks(range(len(CONDITIONS)), list(CONDITIONS))
ax.set_ylim(len(CONDITIONS) - 0.5, -0.75)
ax.text(0.18, -0.55, "← less severe with condition", fontsize=8, color="#555")
ax.text(4.8, -0.55, "more severe with condition →", fontsize=8, color="#555", ha="right")
ax.set(xlabel="Ratio of high-impact shares: Severity 3–4 % with the condition ÷ without (log scale; 1 = no effect)",
       title="RQ1–RQ3 stratified by data provider")
ax.legend(loc="upper left", bbox_to_anchor=(1.01, 1), fontsize=8)
save(fig, "15_rq_by_source")
```

**Insight:** stratifying changes three conclusions.
- **Night (RQ1):** pooled, night looks very slightly *less* high-impact (×0.96), but inside both large providers it is slightly *more* (×1.08, ×1.10): the pooled number was a provider-mix effect. Severity 4 is more common at night in both (Source1 5.2% vs 3.9%, Source2 0.65% vs 0.36%).
- **Traffic signals (RQ3) reverse direction:** ×0.23 in Source2 but ×1.22 in Source1. The pooled "signals → less severe" result (×0.45) comes from how Source2 labels city-street accidents (**Simpson's paradox**).
- **Highway** stays the clearest road signal but its size depends on the provider (×3.6 Source2, ×1.5 Source1); **junctions** are the most stable effect (×1.65–1.78 everywhere), stronger inside providers than pooled.

Weather (RQ2) stays weak in every provider (V ≤ 0.04). Consequence: the `Source` ablation in the modelling phase is necessary, not optional.

## 8. Correlations: Spearman and Cramér's V

Severity is **ordinal** (1 < 2 < 3 < 4) and many numeric columns are skewed, so we use the **Spearman** rank correlation (only assumes a monotonic link, not a straight line) instead of Pearson. For categorical variables we use a **Cramér's V** matrix.

```python
df["road_feature_count"] = df[["Amenity", "Bump", "Crossing", "Give_Way", "Junction", "No_Exit", "Railway", "Roundabout",
                               "Station", "Stop", "Traffic_Calming", "Traffic_Signal"]].sum(axis=1)
df["civil_night"] = df["Civil_Twilight"].astype(str).eq("Night").astype("int8")
SPEARMAN = ["Severity", "Start_Lat", "Start_Lng", "Temperature(F)", "Humidity(%)", "Pressure(in)", "Visibility(mi)",
            "Wind_Speed(mph)", "Precipitation(in)", "hour", "month", "year", "road_feature_count", "is_highway",
            "Traffic_Signal", "Junction", "civil_night", "bad_weather"]
sp = df[SPEARMAN].corr(method="spearman")
fig, ax = plt.subplots(figsize=(11, 9))
sns.heatmap(sp, annot=True, fmt=".2f", annot_kws={"size": 7}, cmap="RdBu_r", center=0, vmin=-1, vmax=1, square=True,
            ax=ax, cbar_kws={"label": "Spearman ρ", "shrink": 0.8})
ax.set(title="Spearman rank correlation (Severity treated as ordinal)")
save(fig, "10_spearman_heatmap")
sp["Severity"].drop("Severity").sort_values(key=abs, ascending=False).round(3).to_frame("spearman_with_Severity")
```

```python
CRAMER = ["Severity", "Source", "year", "State", "season", "Sunrise_Sunset", "Weather_Group", "Wind_Direction",
          "is_highway", "Traffic_Signal", "Junction", "Crossing", "Stop", "is_rush_hour", "is_weekend"]
SEASON = {12: "Winter", 1: "Winter", 2: "Winter", 3: "Spring", 4: "Spring", 5: "Spring",
          6: "Summer", 7: "Summer", 8: "Summer", 9: "Autumn", 10: "Autumn", 11: "Autumn"}
df["season"] = df["month"].map(SEASON)
df["is_weekend"] = (df["weekday"] >= 5).astype("int8")
df["is_rush_hour"] = ((df["weekday"] < 5) & (df["hour"].between(7, 9) | df["hour"].between(16, 19))).astype("int8")
cv = pd.DataFrame(np.eye(len(CRAMER)), index=CRAMER, columns=CRAMER)
for i, a in enumerate(CRAMER):
    for b in CRAMER[i + 1:]:
        cv.loc[a, b] = cv.loc[b, a] = cramers_v(df[a].astype(str), df[b].astype(str))
fig, ax = plt.subplots(figsize=(11, 9))
sns.heatmap(cv, annot=True, fmt=".2f", annot_kws={"size": 7}, cmap="Reds", vmin=0, vmax=1, square=True, ax=ax,
            cbar_kws={"label": "Cramér's V", "shrink": 0.8})
ax.set(title="Cramér's V between categorical variables (0 = no link, 1 = perfect)")
save(fig, "16_cramers_v_matrix")
cv["Severity"].drop("Severity").sort_values(ascending=False).round(3).to_frame("cramers_v_with_Severity")
```

**Insight:** with Severity, all weather columns have |ρ| < 0.06; the strongest rank correlations are **year (ρ = −0.28)**, is_highway (+0.21), road_feature_count (−0.12) and Traffic_Signal (−0.11) — year being the strongest is another sign of label drift. In the Cramér's V matrix, Severity is most linked to Source (0.29), is_highway (0.22) and year (0.19). **Source and year are linked (V = 0.38)**: year partly measures which provider recorded the accident, so the modelling phase also tests a set without year. Among inputs, Traffic_Signal–road_feature_count (ρ = 0.68) and Crossing–Traffic_Signal (V = 0.48) are the strongest pairs; none passes the 0.85 redundancy limit.

## 10. Statistical hypothesis tests

The charts suggest relationships; the tests check they are not just chance. Significance level **α = 0.05**.

**Which test and why**
- Two **categorical** variables (e.g. Severity vs day/night) → **chi-square test of independence**.
  H0: the two variables are independent. Assumption: every expected count ≥ 5 (checked: `min_expected`).
- A **numeric** variable across the 4 Severity groups (temperature) → **Kruskal-Wallis test**.
  H0: the 4 groups have the same distribution. Chosen instead of ANOVA because temperature is not normally
  distributed and the group variances differ (both checked below), and Kruskal-Wallis does not need those assumptions.

**Big-sample warning:** with ~500,000 rows, even tiny differences give p ≈ 0. So each test also reports an
**effect size** (how big the difference is) and is repeated on a random **5,000-row sample**:
- Cramér's V (chi-square): Cohen's 0.1 small, 0.3 medium, 0.5 large hold for 2-value variables; in general they are
  divided by √df\*, with df\* = min(rows − 1, columns − 1) of the table (Source: df\* = 2 → 0.07 / 0.21 / 0.35;
  weather: df\* = 3 → 0.06 / 0.17 / 0.29).
- ε² (Kruskal-Wallis): 0.01 small, 0.06 medium, 0.14 large.

**Multiple tests:** 6 tests are run, so the 5,000-row check uses the Bonferroni level α′ = 0.05 / 6 ≈ 0.008.

```python
ALPHA_BONF = 0.05 / 6  # 6 hypothesis tests


def effect(v, cuts=(0.1, 0.3, 0.5)):
    return "negligible" if v < cuts[0] else "small" if v < cuts[1] else "medium" if v < cuts[2] else "large"


def chi2_test(name, a, b, sample=None):
    ct = pd.crosstab(a, b)
    chi2, p, dof, exp = stats.chi2_contingency(ct)
    df_star = min(ct.shape) - 1                       # Cohen's thresholds scale with 1/sqrt(df*)
    v = np.sqrt(chi2 / (ct.to_numpy().sum() * df_star))
    p5k = stats.chi2_contingency(pd.crosstab(a.loc[sample], b.loc[sample]))[1]
    return {"H0: Severity independent of": name, "chi2": round(chi2), "dof": dof, "p_value": p,
            "min_expected": round(exp.min(), 1), "cramers_v": round(v, 3), "df_star": df_star,
            "effect": effect(v, tuple(w / np.sqrt(df_star) for w in (0.1, 0.3, 0.5))),
            "p_value_5k_sample": p5k, "reject_H0": p < 0.05, "confirmed_5k_bonferroni": p5k < ALPHA_BONF}


s5k = df.sample(5_000, random_state=42).index
known = df["Weather_Group"].ne("Other")  # "Other" (203 rows) gives expected counts < 5 -> excluded
tests = pd.DataFrame([
    chi2_test("data provider (Source)", df["Source"], df["Severity"], s5k),
    chi2_test("day / night", df["Sunrise_Sunset"], df["Severity"], s5k),
    chi2_test("traffic signal nearby", df["Traffic_Signal"], df["Severity"], s5k),
    chi2_test("junction nearby", df["Junction"], df["Severity"], s5k),
    chi2_test("weather group (without 'Other')", df.loc[known, "Weather_Group"].astype(str),
              df.loc[known, "Severity"], s5k.intersection(df.index[known])),
])
pd.set_option("display.float_format", lambda x: f"{x:.3g}")
tests
```

```python
# Assumption checks for temperature (numeric across 4 groups)
temp = [df.loc[df["Severity"] == s, "Temperature(F)"].dropna() for s in (1, 2, 3, 4)]
print("Normality (D'Agostino, 5,000-row sample): p =", f"{stats.normaltest(temp[1].sample(5000, random_state=42)).pvalue:.2g}",
      "-> not normal" )
print("Equal variances (Levene): p =", f"{stats.levene(*temp).pvalue:.2g}", "-> variances differ")

H, p = stats.kruskal(*temp)
n = sum(len(g) for g in temp)
eps2 = (H - len(temp) + 1) / (n - len(temp))
t5k = df.loc[s5k]
p5k = stats.kruskal(*[t5k.loc[t5k["Severity"] == s, "Temperature(F)"].dropna() for s in (1, 2, 3, 4)]).pvalue
pd.DataFrame([{"H0": "Temperature is the same for all Severity levels", "H": round(H, 1), "p_value": p,
               "epsilon_squared": round(eps2, 4), "effect": effect(eps2, (0.01, 0.06, 0.14)),
               "p_value_5k_sample": p5k, "reject_H0": p < 0.05,
               "medians_by_severity": [round(g.median(), 1) for g in temp]}])
```

**Interpretation**
- **Data provider (Source):** H0 rejected; the strongest relationship tested. With df* = 2 the thresholds are
  0.07 / 0.21 / 0.35, so V = 0.29 is a **medium** effect, and it is confirmed on 5,000 rows (p ≈ 10^-182).
  Severity depends a lot on *who reported* the accident.
- **Traffic signal nearby:** H0 rejected, **small** effect (V = 0.12), confirmed on 5,000 rows. Section 7b shows this
  pooled effect depends on the provider (it reverses in Source1).
- **Day / night:** H0 rejected and confirmed on 5,000 rows (p = 0.0004 < 0.008), but the effect is **negligible** (V = 0.05).
- **Weather group** (df* = 3, thresholds 0.06 / 0.17 / 0.29) and **junction nearby:** **negligible** effects
  (V = 0.04 and 0.05) and **not confirmed** on 5,000 rows at the Bonferroni level (p = 0.013 and 0.13).
- **Temperature (Kruskal-Wallis):** H0 rejected, even on 5,000 rows, but ε² = 0.004 is **negligible**:
  Severity 4 accidents are only a little colder (median 60 °F vs 64 °F).

**Conclusion:** with a huge sample, almost everything is "significant", so effect sizes decide what matters.
Only the provider (medium) and traffic signals (small) pass our "meaningful" bar. No single variable explains
severity on its own, which supports using many features together in a non-linear model.

## Key findings

1. **Strong imbalance:** 80% Severity 2, 2.6% Severity 4 → use macro F1, class weights, per-class recall.
2. **Provider bias is the biggest pattern in the data:** Source2 records far more Severity 3, Source1 far more Severity 4,
   and the provider mix changes by year (V(Source, year) = 0.38) → severity shifts over time. `Source` will likely be a top
   feature; its ablation (and a set without `year`) is essential.
3. **Night > rush hour:** Severity 4 is highest around midnight (5–6%) and lowest at morning rush (1.6%), also inside each provider.
4. **Road context matters, but stratify:** highways and junctions raise the high-impact share in every provider; the pooled
   "signals → less severe" effect reverses in Source1 (Simpson's paradox).
5. **Weather matters only a little:** rain/storms raise the Severity 3–4 share by ~5 points pooled; V ≤ 0.04 in every provider.
6. **Weak, non-linear signal:** all weather |ρ| < 0.06 with Severity → expect tree models (Random Forest, LightGBM) to beat
   Logistic Regression.
7. **Hypothesis tests confirm it:** all relationships are significant on the full data, but effect sizes rank them:
   provider (V = 0.29, medium) > traffic signal (0.12, small) > day/night, junction, weather (negligible); temperature ε² = 0.004.
